```python
class Human:

    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 大人判定メソッド
    def check_adult(self):
        if self.age >= 20:
            print(self.name + "さんは大人です。")
        else:
            print(self.name + "さんは大人ではありません。")

# インスタンス生成
human1 = Human("山田太郎", 25)
human2 = Human("佐藤花子", 18)
human3 = Human("田中一郎", 20)

# リストに追加
human_list = [human1, human2, human3]

# 繰り返し処理
for human in human_list:
    human.check_adult()
```
