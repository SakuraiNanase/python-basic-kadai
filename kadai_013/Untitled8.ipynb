```python
# 消費税を含めた金額を計算する関数を定義します
# price：商品の金額
# tax_rate：消費税率（％）
def calculate_tax(price, tax_rate):

    # 消費税を計算します
    # 例：110円 × 10 ÷ 100 = 11円
    tax_amount = price * tax_rate / 100

    # 税込みの合計金額を計算します
    # 商品金額 + 消費税
    total = price + tax_amount

    # 計算した税込み金額を戻り値として返します
    return total


# 関数を呼び出します
# 第1引数：110（商品金額）
# 第2引数：10（消費税率）
result = calculate_tax(110, 10)

# 計算結果を表示します
print(result)
```
